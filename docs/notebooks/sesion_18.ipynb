## Comparar categorías con barras

Un gráfico de barras compara magnitudes entre categorías. Ordenar valores facilita encontrar máximos y mínimos. Un eje que parte de cero evita exagerar diferencias de longitud.

Un título útil expresa el contenido y período. Las etiquetas deben incluir unidades. No dependas solo del color para distinguir información.

```python
import matplotlib.pyplot as plt

productos = ["Regla", "Lapicero", "Cuaderno"]
ingresos = [1200, 5000, 6000]
fig, ax = plt.subplots(figsize=(7, 4))
ax.barh(productos, ingresos, color="#216556")
ax.set(title="Ingresos de la venta de práctica", xlabel="Ingreso (CRC)")
ax.set_xlim(left=0)
fig.tight_layout()
plt.show()
```

**Qué observar:** Cuaderno tiene la barra más larga (₡6000) y Regla la más corta (₡1200).

**Recuerda:** El gráfico muestra ingreso, no ganancia. Usa nombres precisos para el indicador.

## Distribuciones y relaciones

Un histograma agrupa una variable numérica en intervalos para observar su distribución. El número de intervalos afecta la apariencia; prueba opciones razonables e informa el tamaño de la muestra.

Un diagrama de dispersión representa pares de valores. Sirve para explorar relaciones, grupos y observaciones inusuales. Cada punto debe tener un significado claro.

```python
import matplotlib.pyplot as plt

cantidades = [1, 1, 2, 2, 2, 3, 4, 5, 8]
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(cantidades, bins=[0.5, 2.5, 4.5, 6.5, 8.5], color="#7155ba", edgecolor="white")
ax.set(title="Unidades por registro (n=9)", xlabel="Unidades", ylabel="Registros")
fig.tight_layout()
plt.show()
```

**Qué observar:** Los intervalos contienen 5, 2, 1 y 1 registros respectivamente.

**Recuerda:** Una frecuencia cuenta observaciones; no es automáticamente porcentaje ni densidad.

## Tiempo y exportación legible

Un gráfico de líneas es útil cuando el orden temporal tiene sentido. Ordena las fechas y evita tratar categorías arbitrarias como una serie temporal.

Exporta con suficiente resolución y revisa que títulos y etiquetas no estén cortados. Incluye la fuente o explica si son datos de práctica.

```python
import pandas as pd
import matplotlib.pyplot as plt

fechas = pd.to_datetime(["2026-01-01", "2026-01-02", "2026-01-03"])
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(fechas, [99, 216, 102], marker="o", color="#216556")
ax.set(title="Ingresos diarios · ejemplo didáctico", xlabel="Fecha", ylabel="Ingreso (unidades monetarias)")
fig.autofmt_xdate()
fig.tight_layout()
fig.savefig("ingresos_diarios.png", dpi=150, bbox_inches="tight")
plt.show()
```

**Qué observar:** El segundo día tiene el ingreso más alto. Se guarda ingresos_diarios.png.

**Recuerda:** Tres días no bastan para afirmar una tendencia estable ni un patrón estacional.
