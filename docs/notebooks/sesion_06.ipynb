## loc e iloc sobre un índice explícito

loc busca etiquetas; iloc posiciones desde cero. Con etiquetas g1 a g4 ordenadas, loc["g3"] e iloc[2] coinciden.

```python
import pandas as pd
```

```python
equipo = pd.DataFrame({'Nombre': ['Ana', 'Luis', 'Eva', 'Leo'], 'Horas': [2, 4, 3, 5]}, index=['g1', 'g2', 'g3', 'g4'])
```

```python
equipo.loc['g3']
```

```python
equipo.iloc[2]
```

**Interpretación:** Ambas selecciones devuelven a Eva con tres horas.

## Rangos y columnas

En un índice único y ordenado, loc incluye la etiqueta final. iloc excluye el extremo final. En el CSV real verifica los nombres antes de seleccionar las posiciones 1, 3, 11 y 25; las posiciones de esta muestra son distintas.

```python
import pandas as pd
```

```python
peliculas = pd.DataFrame({'movie_title': [' Aurora ', 'Brisa', 'Cauce', 'Delta', 'Eco'], 'director_name': ['Ana', 'Luis', 'Ana', 'Eva', 'Luis'], 'duration': [160.0, 95.0, None, 130.0, 170.0], 'imdb_score': [8.2, 7.1, 8.2, 6.8, 8.7], 'title_year': [2005, 1998, 2010, 2020, 1995], 'language': [' Spanish ', 'English', 'Spanish', None, 'English'], 'country': ['Spain', 'USA', 'Spain', None, 'USA'], 'num_voted_users': [100, 200, 300, 50, 400]})
```

```python
peliculas.iloc[:3, [1, 2, 0, 3]]
```

```python
peliculas.columns[[1, 2, 0, 3]]
```

```python
peliculas.loc[0:2, ['movie_title', 'imdb_score']]
```

**Interpretación:** Se seleccionan las primeras tres filas; consulta los encabezados para verificar las columnas.

## El índice después de ordenar

sort_values cambia el orden de las filas, no sus etiquetas. Sin reset_index, loc[0] sigue buscando el registro con etiqueta 0; iloc[0] busca la primera posición actual.

```python
import pandas as pd
```

```python
peliculas = pd.DataFrame({'movie_title': [' Aurora ', 'Brisa', 'Cauce', 'Delta', 'Eco'], 'director_name': ['Ana', 'Luis', 'Ana', 'Eva', 'Luis'], 'duration': [160.0, 95.0, None, 130.0, 170.0], 'imdb_score': [8.2, 7.1, 8.2, 6.8, 8.7], 'title_year': [2005, 1998, 2010, 2020, 1995], 'language': [' Spanish ', 'English', 'Spanish', None, 'English'], 'country': ['Spain', 'USA', 'Spain', None, 'USA'], 'num_voted_users': [100, 200, 300, 50, 400]})
```

```python
ordenadas = peliculas.sort_values('imdb_score', ascending=False)
```

```python
ordenadas.iloc[0][['movie_title', 'imdb_score']]
```

```python
ordenadas.loc[0, ['movie_title', 'imdb_score']]
```

**Interpretación:** Primera posición: Eco, 8.7. Etiqueta 0: Aurora, 8.2.
