## Series y DataFrames

Una Series relaciona valores con etiquetas. Un DataFrame reúne columnas con un índice común. Seleccionar un nombre devuelve una Series; una lista de nombres conserva un DataFrame.

```python
import pandas as pd
```

```python
peliculas = pd.DataFrame({'movie_title': [' Aurora ', 'Brisa', 'Cauce', 'Delta', 'Eco'], 'director_name': ['Ana', 'Luis', 'Ana', 'Eva', 'Luis'], 'duration': [160.0, 95.0, None, 130.0, 170.0], 'imdb_score': [8.2, 7.1, 8.2, 6.8, 8.7], 'title_year': [2005, 1998, 2010, 2020, 1995], 'language': [' Spanish ', 'English', 'Spanish', None, 'English'], 'country': ['Spain', 'USA', 'Spain', None, 'USA'], 'num_voted_users': [100, 200, 300, 50, 400]})
```

```python
puntuaciones = pd.Series([7.4, 8.1, 6.9], index=['Luna', 'Mar', 'Sol'])
```

```python
puntuaciones
```

```python
peliculas[['movie_title', 'imdb_score']].head()
```

**Interpretación:** Las etiquetas de la Series nombran películas. La tabla conserva dos columnas.
