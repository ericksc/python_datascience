## Series y DataFrame

Una Series combina valores con un índice. Un DataFrame es una tabla de columnas que comparten un índice de filas. Los nombres de columnas describen variables y cada fila representa una observación.

Al construir una tabla con listas, cada columna debe tener la misma longitud. El significado de una fila depende del conjunto de datos: producto, persona o registro de venta.

```python
import pandas as pd

productos = pd.DataFrame({
    "Producto": ["Mouse", "Teclado", "Monitor"],
    "Precio": [15.99, 25.50, 199.99],
    "Cantidad": [3, 2, 1]
}, index=["p1", "p2", "p3"])
print(productos)
print(type(productos["Precio"]).__name__)
```

**Qué observar:** Una tabla con tres filas y tres columnas; seleccionar Precio devuelve una Series.

**Recuerda:** El índice identifica filas en la tabla. No necesariamente representa un identificador de negocio.

## Inspeccionar antes de operar

head muestra las primeras filas; shape devuelve dimensiones; columns contiene nombres; info resume tipos y valores no nulos. Una vista inicial ayuda a detectar problemas antes de sumar.

Una selección con un nombre devuelve una Series. Una selección con una lista de nombres devuelve un DataFrame, incluso si esa lista contiene un único nombre.

```python
import pandas as pd

productos = pd.DataFrame({
    "Producto": ["Mouse", "Teclado", "Monitor"],
    "Precio": [15.99, 25.50, 199.99],
    "Cantidad": [3, 2, 1]
}, index=["p1", "p2", "p3"])
print(productos.head(2))
print(productos.shape)
print(productos.columns.tolist())
productos.info()
print(productos[["Producto", "Precio"]])
```

**Qué observar:** Forma (3, 3). Precio es numérico decimal; Cantidad es entera.

**Recuerda:** shape y columns son atributos, sin paréntesis. head e info son métodos, con paréntesis.

## Leer y exportar CSV

read_csv interpreta un archivo y crea una tabla. Revisa siempre el separador, las columnas y los tipos. El archivo ventas.csv del curso contiene cinco registros; su moneda no está especificada.

Al exportar, index=False evita guardar el índice como una columna adicional. Para Excel se usa read_excel o to_excel y se requiere un motor compatible, como openpyxl para xlsx.

```python
import pandas as pd
from pathlib import Path

DATOS = next((p for p in [Path("datos"), Path("../datos"), Path(".")]
              if (p / "ventas.csv").exists()), Path("datos"))
ventas = pd.read_csv(DATOS / "ventas.csv")
print(ventas.shape)
ventas.to_csv("ventas_copia.csv", index=False)
print("Copia exportada")
```

**Qué observar:** (5, 4) y un archivo ventas_copia.csv en la carpeta de trabajo.

**Recuerda:** Conserva el archivo original. Exporta a un nombre nuevo para mantener una referencia verificable.
