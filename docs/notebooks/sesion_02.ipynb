## Listas: una secuencia con orden

Una lista reúne valores en un orden. Los índices comienzan en cero y -1 selecciona el último elemento. len informa cuántos elementos contiene.

El rango [inicio:fin] incluye el inicio y excluye el final. append agrega un elemento al final y modifica la lista existente.

```python
productos = ["Cuaderno", "Lapicero", "Regla"]
print(productos[0])
print(productos[-1])
print(productos[0:2])
productos.append("Carpeta")
print(len(productos))
```

**Qué observar:** Cuaderno; Regla; [Cuaderno, Lapicero]; 4.

**Recuerda:** No confundas cantidad de elementos con última posición: una lista de cuatro elementos termina en la posición 3.

## Diccionarios: un registro con nombres

Un diccionario relaciona claves con valores. Es una buena forma de representar un registro: Producto, Precio y Cantidad explican qué significa cada valor.

Acceder a una clave inexistente con corchetes produce KeyError. get permite expresar un valor por defecto cuando la ausencia es esperable. No ocultes errores tipográficos con valores por defecto.

```python
venta = {"Producto": "Cuaderno", "Precio": 1500, "Cantidad": 4}
print(venta["Producto"])
venta["Ingreso"] = venta["Precio"] * venta["Cantidad"]
print(venta["Ingreso"])
print(venta.get("Descuento", 0))
```

**Qué observar:** Cuaderno; 6000; 0.

**Recuerda:** Las claves son nombres; las posiciones son números que describen un lugar. Son dos formas distintas de acceder a datos.

## Elegir la colección adecuada

Una lista de diccionarios puede representar varias ventas. Una tupla agrupa valores que no se modifican mediante asignación de elementos. Un conjunto guarda elementos únicos y no ofrece acceso por posición.

Convertir una lista en conjunto sirve para reconocer categorías únicas. No utilices el orden de un conjunto para organizar una entrega; ordénalo explícitamente si lo necesitas.

```python
ventas = [{"Producto": "Mouse", "Cantidad": 3},
          {"Producto": "Teclado", "Cantidad": 2}]
print(ventas[1]["Producto"])
coordenada = (10, 20)
print(coordenada[0])
ciudades = ["San José", "Cartago", "San José"]
print(sorted(set(ciudades)))
```

**Qué observar:** Teclado; 10; [Cartago, San José].

**Recuerda:** Eliminar repeticiones no siempre es limpiar: dos ventas iguales pueden ser eventos válidos distintos.
