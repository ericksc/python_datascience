## Una pregunta antes del código

Una librería vende cuatro cuadernos a ₡1500 cada uno. Define primero la pregunta: ¿cuál es el ingreso de esa venta? El precio es un monto por unidad y la cantidad es un número de unidades. Su producto es un monto total.

En un notebook conviven celdas de texto y código. Ejecuta con Shift + Enter. Python conserva variables mientras la sesión está activa: una celda que usa precio necesita que antes se haya ejecutado su asignación.

```python
precio = 1500
cantidad = 4
ingreso = precio * cantidad
print("Ingreso:", ingreso)
```

**Qué observar:** Ingreso: 6000

**Recuerda:** ₡6000 es ingreso por ventas. No es ganancia: todavía no has descontado costos.

## Un nombre para cada dato

Una variable asocia un nombre con un valor. Usa nombres como precio_unitario en lugar de letras ambiguas. Python distingue mayúsculas: Precio y precio son nombres diferentes.

str representa texto; int, enteros; float, decimales; bool, verdadero o falso. El signo = asigna un valor. El operador == compara dos valores.

```python
producto = "Cuaderno"
cantidad = 4
precio = 1500.0
hay_stock = cantidad > 0
print(type(producto).__name__)
print(type(cantidad).__name__)
print(type(precio).__name__)
print(hay_stock)
```

**Qué observar:** str, int, float y True, en líneas separadas.

**Recuerda:** Escribe texto entre comillas. True y False se escriben con inicial mayúscula y sin comillas.

## Convertir, calcular y comprobar

Los datos pueden llegar como texto. float("1500.50") convierte un texto numérico a decimal; int("4") lo convierte a entero. Una conversión falla si el texto no representa el tipo esperado.

Lee un mensaje de error desde su última línea: suele indicar el tipo y la causa. Reiniciar y ejecutar todo permite detectar dependencias ocultas entre celdas.

```python
precio_texto = "1500.50"
cantidad_texto = "4"
total = float(precio_texto) * int(cantidad_texto)
print(round(total, 2))
print(total > 5000)
```

**Qué observar:** 6002.0 y True.

**Recuerda:** Los decimales binarios pueden ser aproximados. Redondear al mostrar es útil; los cálculos financieros exactos requieren decisiones adicionales.
