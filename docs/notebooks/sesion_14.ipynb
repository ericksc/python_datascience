## De una lista a un arreglo

Un arreglo de NumPy organiza valores de un tipo compatible y permite operaciones numéricas sobre todos sus elementos. Multiplicar una lista por dos repite sus elementos; multiplicar un arreglo por dos duplica cada valor.

La vectorización expresa una operación para el conjunto, sin escribir un ciclo explícito. Esto facilita leer la intención del cálculo.

```python
import numpy as np

lista = [2, 4, 6]
arreglo = np.array(lista)
print(lista * 2)
print(arreglo * 2)
print(arreglo.mean())
```

**Qué observar:** La lista se repite: [2, 4, 6, 2, 4, 6]. El arreglo produce [4, 8, 12]. Su media es 4.0.

**Recuerda:** No asumas que dos objetos responden igual al mismo operador: revisa su tipo.

## Forma, posiciones y ejes

shape indica cuántos elementos hay en cada dimensión. Una matriz de 2 × 3 tiene dos filas y tres columnas. El acceso [fila, columna] usa posiciones desde cero.

En una tabla bidimensional, sum(axis=0) suma a través de las filas y devuelve un total por columna. sum(axis=1) suma las columnas y devuelve un total por fila.

```python
import numpy as np

matriz = np.array([[2, 4, 6], [1, 3, 5]])
print(matriz.shape)
print(matriz[1, 0])
print(matriz.sum(axis=0))
print(matriz.sum(axis=1))
```

**Qué observar:** (2, 3); 1; [3, 7, 11]; [12, 9].

**Recuerda:** Describe la salida con palabras antes de elegir el eje: ¿necesitas un resultado por fila o por columna?

## Máscaras y datos de simulación

Una comparación sobre un arreglo produce valores booleanos. Se pueden usar como máscara para conservar elementos que cumplen una condición.

Un generador con semilla permite repetir una simulación en el mismo entorno. No convierte los datos simulados en evidencia del mundo real. El límite superior de integers es exclusivo.

```python
import numpy as np

precios = np.array([500, 1500, 600, 1200])
print(precios[precios > 700])
rng = np.random.default_rng(42)
simulacion = rng.integers(1, 6, size=5)
print(simulacion)
print(np.all((simulacion >= 1) & (simulacion <= 5)))
```

**Qué observar:** El filtro devuelve [1500, 1200]. La simulación contiene cinco enteros entre 1 y 5; la comprobación es True.

**Recuerda:** Fija la semilla y registra las versiones si necesitas reproducir una simulación con precisión.
