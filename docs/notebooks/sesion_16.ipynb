## Máscara booleana: True selecciona

Una comparación produce un arreglo booleano. x[mask] conserva valores donde la máscara es True; al seleccionar de una matriz puede devolver un vector y perder la organización original.

np.where permite elegir entre alternativas conservando una forma compatible. Para detectar valores finitos usa np.isfinite, que excluye NaN e infinito.

```python
import numpy as np
x=np.array([[1.,np.nan],[3.,np.inf]])
valido=np.isfinite(x)
print(x[valido])
print(np.where(valido,x,0))
```

**Qué observar:** Selección: [1,3]. where conserva 2×2 sustituyendo inválidos por 0 en esta demostración.

**Recuerda:** Sustituir por cero cambia estadísticas. No lo uses como limpieza automática.

## MaskedArray: True excluye

numpy.ma combina valores y una máscara, manteniendo dimensiones. A diferencia de la selección booleana, True en la máscara significa que el valor está excluido. Sus reducciones ignoran los valores enmascarados.

masked_invalid enmascara NaN e infinitos. También puedes definir reglas de dominio, por ejemplo temperaturas de sensor fuera de rango. La máscara expresa tu decisión y no inventa mediciones.

```python
import numpy as np
x=np.array([10.,20.,-999.,30.])
m=np.ma.masked_where(x==-999,x)
print(m)
print(m.mean())
print(m.count())
```

**Qué observar:** Valor -999 enmascarado; media 20.0; conteo válido 3.

**Recuerda:** Recuerda la inversión: True selecciona en x[mask], pero True excluye en numpy.ma.

## Cobertura y grupos totalmente inválidos

count informa cuántos valores válidos quedan por eje. Un grupo sin datos válidos debe mantenerse como desconocido; reportar cero sería engañoso.

filled convierte la máscara a un valor de relleno. Con floats puedes usar NaN para exportar. Documenta las reglas y conserva el conteo de válidos junto a cualquier promedio.

```python
import numpy as np
x=np.array([[10.,np.nan],[20.,np.inf]])
m=np.ma.masked_invalid(x)
print(m.count(axis=0))
print(m.mean(axis=0))
print(m.mean(axis=0).filled(np.nan))
```

**Qué observar:** Conteos [2,0]; medias [15,enmascarado]; al exportar [15,NaN].

**Recuerda:** Una media sin cobertura puede ocultar que todos los datos del grupo son inválidos.
