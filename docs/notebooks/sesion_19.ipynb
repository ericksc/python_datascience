## Figure, Axes y paneles con propósito

Figure es el lienzo completo y cada Axes es una región de trazado. Crear ambos explícitamente facilita combinar gráficos y modificar un eje sin alterar otros.

Un panel debe responder a una pregunta. Aquí comparamos volumen e ingreso por producto. No compartimos el eje de valores porque las unidades son distintas; sí mantenemos el mismo orden de categorías.

```python
import pandas as pd
datos = pd.DataFrame({
    "Sucursal": ["Centro"]*6 + ["Norte"]*6,
    "Producto": ["A", "B", "C"]*4,
    "Unidades": [2, 4, 3, 5, 2, 6, 3, 5, 4, 6, 3, 7],
    "Precio": [100, 200, 150]*4
})
datos["Ingreso"] = datos["Unidades"] * datos["Precio"]
import matplotlib.pyplot as plt
r=datos.groupby("Producto")[["Unidades","Ingreso"]].sum()
fig,axes=plt.subplots(1,2,figsize=(10,4),layout="constrained")
axes[0].bar(r.index,r.Unidades,color="#216556")
axes[1].bar(r.index,r.Ingreso,color="#7155ba")
axes[0].set(title="Volumen por producto",ylabel="Unidades")
axes[1].set(title="Ingreso por producto",ylabel="Ingreso (CRC)")
plt.show()
```

**Qué observar:** Dos paneles comparan productos A, B y C con unidades claramente distintas.

**Recuerda:** Compartir escala solo es útil cuando la magnitud y la unidad son comparables.

## Títulos, etiquetas y anotaciones útiles

Un título puede expresar un hallazgo verificable, pero debe limitarse al conjunto observado. Usa anotaciones para señalar un punto que merece atención; no conviertas cada dato en una etiqueta.

Las barras parten de cero para conservar la comparación de longitudes. Usa color con significado, contraste suficiente y etiquetas que permitan comprender el gráfico sin distinguir colores.

```python
import pandas as pd
datos = pd.DataFrame({
    "Sucursal": ["Centro"]*6 + ["Norte"]*6,
    "Producto": ["A", "B", "C"]*4,
    "Unidades": [2, 4, 3, 5, 2, 6, 3, 5, 4, 6, 3, 7],
    "Precio": [100, 200, 150]*4
})
datos["Ingreso"] = datos["Unidades"] * datos["Precio"]
import matplotlib.pyplot as plt
r=datos.groupby("Producto").Ingreso.sum().sort_values()
fig,ax=plt.subplots(figsize=(7,4),layout="constrained")
barras=ax.barh(r.index,r.values,color="#216556")
ax.bar_label(barras,fmt="%.0f",padding=4)
ax.set(xlabel="Ingreso (CRC)",title="Ingreso registrado por producto · datos sintéticos")
ax.set_xlim(0,r.max()*1.2)
ax.spines[["top","right"]].set_visible(False)
plt.show()
```

**Qué observar:** A: 1600; B: 2800; C: 3000 CRC. Los valores aparecen al final de las barras.

**Recuerda:** No llames ganancia a un ingreso: el conjunto no incluye costos.

## Exportar para pantalla y documento

PNG es una imagen de píxeles; SVG conserva formas vectoriales que escalan con nitidez. Elige formato según el destino, no por costumbre. dpi afecta principalmente la resolución de los elementos rasterizados.

Guarda la figura antes de cerrar el entorno. bbox_inches="tight" reduce recortes, pero igualmente revisa título, leyendas, ejes y tamaño real de lectura. Evita títulos diminutos para hacer caber más información.

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(figsize=(7,4),layout="constrained")
ax.plot([1,2,3],[10,15,12],marker="o",color="#216556")
ax.set(title="Unidades registradas en tres días",xlabel="Día",ylabel="Unidades")
fig.savefig("reporte_matplotlib.png",dpi=150,bbox_inches="tight")
fig.savefig("reporte_matplotlib.svg",bbox_inches="tight")
plt.show()
```

**Qué observar:** Se generan reporte_matplotlib.png y reporte_matplotlib.svg.

**Recuerda:** Una imagen atractiva no sustituye la fuente, las unidades ni una interpretación correcta.
