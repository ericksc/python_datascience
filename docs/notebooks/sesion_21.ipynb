## Plotly Express: de tabla a figura

Plotly Express devuelve una Figure que puede personalizarse después. Los gráficos permiten explorar puntos con el cursor y ocultar series desde la leyenda. En px.bar los valores no se resumen automáticamente como en un barplot estadístico: prepara los totales antes.

hover_data controla qué columnas aparecen al explorar. Un dato oculto en el gráfico puede seguir presente en el archivo exportado; comparte únicamente los datos que corresponde distribuir.

```python
import pandas as pd
datos = pd.DataFrame({
    "Sucursal": ["Centro"]*6 + ["Norte"]*6,
    "Producto": ["A", "B", "C"]*4,
    "Unidades": [2, 4, 3, 5, 2, 6, 3, 5, 4, 6, 3, 7],
    "Precio": [100, 200, 150]*4
})
datos["Ingreso"] = datos["Unidades"] * datos["Precio"]
import plotly.express as px
r=datos.groupby(["Sucursal","Producto"],as_index=False).Ingreso.sum()
fig=px.bar(r,x="Producto",y="Ingreso",color="Sucursal",barmode="group",labels={"Ingreso":"Ingreso (CRC)"},title="Ingreso por producto y sucursal")
fig.update_layout(template="plotly_white",legend_title_text="Sucursal")
fig.show()
```

**Qué observar:** Barras agrupadas para Centro y Norte. La leyenda permite ocultar series y el cursor muestra sus valores.

**Recuerda:** La interacción debe ayudar a responder una pregunta; no reemplaza título, unidades ni contexto.

## Trazas, layout y exploración temporal

Una figura separa datos y apariencia: las trazas contienen series; layout define títulos, ejes, leyendas y otras opciones. update_traces modifica propiedades de las series y update_layout modifica la composición.

Ordena por fecha antes de unir puntos con líneas. Un selector de rango permite explorar un intervalo sin cambiar los datos originales. Aclara que un cambio de escala o rango altera lo visible, no los valores almacenados.

```python
import pandas as pd
import plotly.express as px
df=pd.DataFrame({"Fecha":pd.date_range("2026-01-01",periods=8),"Ingreso":[100,120,90,150,160,140,180,170]})
fig=px.line(df.sort_values("Fecha"),x="Fecha",y="Ingreso",markers=True,title="Ingresos diarios · datos sintéticos",labels={"Ingreso":"Ingreso (CRC)"})
fig.update_traces(line_color="#216556",hovertemplate="%{x|%d/%m/%Y}<br>CRC %{y:.0f}<extra></extra>")
fig.update_xaxes(rangeslider_visible=True)
fig.update_layout(template="plotly_white")
fig.show()
```

**Qué observar:** Una serie de ocho días con selector de rango e información de fecha y monto al pasar el cursor.

**Recuerda:** Un zoom no es un filtro analítico aplicado al conjunto; distingue exploración visual y transformación de datos.

## Exporta un HTML que se pueda abrir

write_html permite compartir un gráfico sin un servidor Python. include_plotlyjs=True incluye la biblioteca dentro del HTML: el archivo es más grande, pero puede abrirse sin conexión. Con "cdn", el lector necesita internet para cargarla.

Para PNG o SVG, Plotly requiere herramientas adicionales de exportación como Kaleido y un navegador compatible según la versión. En este curso usamos HTML para evitar ese requisito; abre el resultado y comprueba la interacción antes de entregarlo.

```python
import pandas as pd
datos = pd.DataFrame({
    "Sucursal": ["Centro"]*6 + ["Norte"]*6,
    "Producto": ["A", "B", "C"]*4,
    "Unidades": [2, 4, 3, 5, 2, 6, 3, 5, 4, 6, 3, 7],
    "Precio": [100, 200, 150]*4
})
datos["Ingreso"] = datos["Unidades"] * datos["Precio"]
import plotly.express as px
fig=px.scatter(datos,x="Unidades",y="Ingreso",color="Sucursal",symbol="Producto",hover_data=["Producto"],labels={"Ingreso":"Ingreso (CRC)"},title="Ventas de práctica")
fig.update_layout(template="plotly_white")
fig.write_html("ventas_interactivas.html",include_plotlyjs=True,full_html=True,auto_open=False)
print("Archivo creado: ventas_interactivas.html")
fig.show()
```

**Qué observar:** Se crea ventas_interactivas.html con el gráfico y su biblioteca incluidos.

**Recuerda:** HTML interactivo no equivale a un panel con base de datos ni a actualizaciones en tiempo real.
