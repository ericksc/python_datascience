## Texto, repr y frecuencias

str.strip elimina espacios en extremos; str.lower normaliza mayúsculas. Crea columnas nuevas. repr hace visibles los espacios originales. Estas operaciones no eliminan duplicados. value_counts omite ausencias por defecto.

```python
import pandas as pd
```

```python
peliculas = pd.DataFrame({'movie_title': [' Aurora ', 'Brisa', 'Cauce', 'Delta', 'Eco'], 'director_name': ['Ana', 'Luis', 'Ana', 'Eva', 'Luis'], 'duration': [160.0, 95.0, None, 130.0, 170.0], 'imdb_score': [8.2, 7.1, 8.2, 6.8, 8.7], 'title_year': [2005, 1998, 2010, 2020, 1995], 'language': [' Spanish ', 'English', 'Spanish', None, 'English'], 'country': ['Spain', 'USA', 'Spain', None, 'USA'], 'num_voted_users': [100, 200, 300, 50, 400]})
```

```python
peliculas['movie_title_limpio'] = peliculas['movie_title'].str.strip()
```

```python
[(repr(a), repr(b)) for a, b in zip(peliculas['movie_title'].head(), peliculas['movie_title_limpio'].head())]
```

```python
peliculas['language_normalizado'] = peliculas['language'].str.lower().str.strip()
```

```python
peliculas['language'].value_counts()
```

```python
peliculas['language_normalizado'].value_counts()
```

**Interpretación:** Tras normalizar hay dos registros spanish y dos english; una ausencia no entra en el conteo.

## Groupby básico: media frente a suma

groupby separa registros por país. mean usa puntuaciones disponibles; sum acumula votos registrados. Por defecto las claves de país ausentes quedan fuera. El ranking de medias no mide por sí solo calidad del cine y la suma de votos no cuenta personas únicas.

```python
import pandas as pd
```

```python
peliculas = pd.DataFrame({'movie_title': [' Aurora ', 'Brisa', 'Cauce', 'Delta', 'Eco'], 'director_name': ['Ana', 'Luis', 'Ana', 'Eva', 'Luis'], 'duration': [160.0, 95.0, None, 130.0, 170.0], 'imdb_score': [8.2, 7.1, 8.2, 6.8, 8.7], 'title_year': [2005, 1998, 2010, 2020, 1995], 'language': [' Spanish ', 'English', 'Spanish', None, 'English'], 'country': ['Spain', 'USA', 'Spain', None, 'USA'], 'num_voted_users': [100, 200, 300, 50, 400]})
```

```python
peliculas.groupby('country')['imdb_score'].mean().sort_values(ascending=False)
```

```python
peliculas.groupby('country')['num_voted_users'].sum().sort_values(ascending=False)
```

**Interpretación:** Spain: media 8.2 y 400 votos. USA: media 7.9 y 600 votos. Mayor media no implica mayor volumen de votos.
