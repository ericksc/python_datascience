## Agrupar para responder una pregunta

groupby divide registros según una o varias claves. Una agregación resume cada grupo con sum, mean, count u otra operación. Elige el resumen según la pregunta.

Sumar cantidades devuelve unidades; contar filas devuelve registros. Son indicadores diferentes, aunque a veces tengan el mismo valor.

```python
import pandas as pd

ventas = pd.DataFrame({"Producto": ["Mouse", "Mouse", "Teclado"],
                      "Cantidad": [3, 1, 2], "Ingreso": [47.97, 15.99, 51.0]})
resumen = ventas.groupby("Producto", as_index=False).agg(
    unidades=("Cantidad", "sum"),
    ingreso=("Ingreso", "sum"),
    registros=("Producto", "size")
)
print(resumen)
```

**Qué observar:** Mouse: 4 unidades, 63.96 de ingreso y 2 registros. Teclado: 2, 51.00 y 1.

**Recuerda:** Un promedio de precios por registro no es lo mismo que el precio promedio ponderado por unidades.

## Unir tablas con una clave

merge incorpora información de otra tabla usando claves compartidas. Una unión left conserva todas las filas de la tabla izquierda y añade coincidencias de la derecha.

Si un catálogo tiene claves repetidas, una unión puede multiplicar filas e inflar totales. validate="many_to_one" exige que la clave de la derecha sea única. indicator permite revisar coincidencias.

```python
import pandas as pd

ventas = pd.DataFrame({"Producto": ["Mouse", "Mouse", "Teclado"], "Cantidad": [3, 1, 2]})
catalogo = pd.DataFrame({"Producto": ["Mouse", "Teclado"], "Categoria": ["Accesorios", "Accesorios"]})
unidas = ventas.merge(catalogo, on="Producto", how="left", validate="many_to_one", indicator=True)
print(unidas)
print(len(ventas), len(unidas))
```

**Qué observar:** Tres filas antes y después; todas las coincidencias son both.

**Recuerda:** Comprueba claves, filas y totales antes y después de una unión. Una tabla que se ve bien puede contener filas multiplicadas.

## Fechas y tablas de resumen

to_datetime convierte texto en fechas. Especificar el formato reduce ambigüedades: 03/04 puede significar días distintos según la convención. errors="coerce" permite detectar fechas inválidas.

dt.to_period("M") obtiene el mes. pivot_table reorganiza datos y agrega valores. fill_value=0 es apropiado aquí solo porque el ejemplo define ausencia de registro como ausencia de ventas registradas.

```python
import pandas as pd

ventas = pd.DataFrame({"Fecha": ["2026-01-05", "2026-01-20", "2026-02-02"],
                      "Producto": ["Mouse", "Teclado", "Mouse"], "Ingreso": [48, 51, 32]})
ventas["Fecha"] = pd.to_datetime(ventas["Fecha"], format="%Y-%m-%d", errors="coerce")
ventas["Mes"] = ventas["Fecha"].dt.to_period("M").astype(str)
tabla = ventas.pivot_table(index="Mes", columns="Producto", values="Ingreso", aggfunc="sum", fill_value=0)
print(tabla)
```

**Qué observar:** Enero: Mouse 48, Teclado 51. Febrero: Mouse 32, Teclado 0.

**Recuerda:** Un período incompleto no debe compararse sin aclaración con otro completo.
