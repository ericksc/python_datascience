## Una condición y su conteo

La comparación produce una máscara booleana. loc combina esa máscara con columnas. len cuenta filas seleccionadas; no representa personas ni títulos únicos.

```python
import pandas as pd
```

```python
peliculas = pd.DataFrame({'movie_title': [' Aurora ', 'Brisa', 'Cauce', 'Delta', 'Eco'], 'director_name': ['Ana', 'Luis', 'Ana', 'Eva', 'Luis'], 'duration': [160.0, 95.0, None, 130.0, 170.0], 'imdb_score': [8.2, 7.1, 8.2, 6.8, 8.7], 'title_year': [2005, 1998, 2010, 2020, 1995], 'language': [' Spanish ', 'English', 'Spanish', None, 'English'], 'country': ['Spain', 'USA', 'Spain', None, 'USA'], 'num_voted_users': [100, 200, 300, 50, 400]})
```

```python
seleccion = peliculas.loc[peliculas['duration'] > 150, ['movie_title', 'duration', 'imdb_score']]
```

```python
seleccion
```

```python
len(seleccion)
```

**Interpretación:** Dos registros superan 150 minutos; la duración ausente no cumple la condición.

## Combinar condiciones con y y o

```python
import pandas as pd
```

```python
peliculas = pd.DataFrame({'movie_title': [' Aurora ', 'Brisa', 'Cauce', 'Delta', 'Eco'], 'director_name': ['Ana', 'Luis', 'Ana', 'Eva', 'Luis'], 'duration': [160.0, 95.0, None, 130.0, 170.0], 'imdb_score': [8.2, 7.1, 8.2, 6.8, 8.7], 'title_year': [2005, 1998, 2010, 2020, 1995], 'language': [' Spanish ', 'English', 'Spanish', None, 'English'], 'country': ['Spain', 'USA', 'Spain', None, 'USA'], 'num_voted_users': [100, 200, 300, 50, 400]})
```

```python
recientes = peliculas.loc[(peliculas['imdb_score'] >= 8) & (peliculas['title_year'] >= 2000)]
```

```python
recientes[['movie_title', 'title_year', 'imdb_score']]
```

```python
peliculas.loc[(peliculas['language'] == 'Spanish') | (peliculas['country'] == 'Costa Rica'), ['movie_title', 'language', 'country']]
```

**Interpretación:** Aurora y Cauce cumplen el filtro de puntuación y año. La comparación de texto es exacta: los espacios importan.

## Ordenar con dos criterios

El segundo criterio resuelve empates del primero. ascending acepta una lista para ordenar cada columna en un sentido. Filtra primero, ordena después y finalmente aplica head. Conserva las filas repetidas.

```python
import pandas as pd
```

```python
peliculas = pd.DataFrame({'movie_title': [' Aurora ', 'Brisa', 'Cauce', 'Delta', 'Eco'], 'director_name': ['Ana', 'Luis', 'Ana', 'Eva', 'Luis'], 'duration': [160.0, 95.0, None, 130.0, 170.0], 'imdb_score': [8.2, 7.1, 8.2, 6.8, 8.7], 'title_year': [2005, 1998, 2010, 2020, 1995], 'language': [' Spanish ', 'English', 'Spanish', None, 'English'], 'country': ['Spain', 'USA', 'Spain', None, 'USA'], 'num_voted_users': [100, 200, 300, 50, 400]})
```

```python
peliculas.sort_values(['imdb_score', 'num_voted_users'], ascending=[False, False])[['movie_title', 'imdb_score', 'num_voted_users']].head(3)
```

```python
seleccion = peliculas.loc[(peliculas['imdb_score'] >= 8) & (peliculas['title_year'] >= 2000)]
```

```python
seleccion.sort_values(['imdb_score', 'duration'], ascending=[False, True])[['movie_title', 'duration', 'imdb_score']].head(5)
```

**Interpretación:** Eco aparece primero; Cauce precede a Aurora en el empate por puntuación porque tiene más votos.
