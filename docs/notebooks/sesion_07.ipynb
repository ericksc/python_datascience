## Ausencias y duplicados no son lo mismo

Un valor faltante significa que no tenemos un dato. Un duplicado puede ser un registro cargado dos veces, pero también dos eventos válidos parecidos. Decide usando una clave de negocio.

Antes de borrar, cuenta y revisa. En este ejemplo id identifica la venta; la segunda aparición de id 2 es una copia exacta del mismo registro.

```python
import pandas as pd

datos = pd.DataFrame({"id": [1, 2, 2, 3], "Cantidad": [4, 2, 2, None]})
print(datos.isna().sum())
print("Duplicados exactos:", datos.duplicated().sum())
limpios = datos.drop_duplicates().copy()
print(limpios)
```

**Qué observar:** Cantidad tiene 1 faltante. Hay 1 duplicado exacto. Quedan tres filas.

**Recuerda:** No rellenes todas las ausencias con cero: una cantidad desconocida no equivale a una venta de cero unidades.

## Texto consistente, categorías útiles

Mayúsculas y espacios pueden dividir una misma categoría en varias. str.strip elimina espacios en los extremos; str.lower convierte a minúsculas. Conserva el texto original en otra columna.

Normalizar no resuelve todos los casos. Abreviaturas, errores ortográficos y categorías ambiguas requieren reglas explícitas y revisión.

```python
import pandas as pd

encuesta = pd.DataFrame({"respuesta": ["Buena", " buena ", "MALA", "mala"]})
encuesta["respuesta_limpia"] = encuesta["respuesta"].str.strip().str.lower()
print(encuesta["respuesta_limpia"].value_counts())
```

**Qué observar:** buena: 2; mala: 2.

**Recuerda:** Documenta las reglas aplicadas. Cambiar el texto sin conservar el original dificulta revisar decisiones.

## Tipos y rangos: convertir también revela problemas

to_numeric con errors="coerce" convierte entradas inválidas en faltantes. Eso permite detectarlas, pero no corrige su significado. La cantidad de nuevas ausencias es parte del informe de calidad.

Un valor atípico puede ser un error o un caso real importante. Primero investiga la unidad, la fuente y el contexto. No lo elimines solo por ser grande.

```python
import pandas as pd

datos = pd.DataFrame({"Cantidad_original": ["4", "sin dato", "-2", "100"]})
datos["Cantidad"] = pd.to_numeric(datos["Cantidad_original"], errors="coerce")
print(datos)
print("Inválidos o ausentes:", datos["Cantidad"].isna().sum())
print("Negativos:", (datos["Cantidad"] < 0).sum())
```

**Qué observar:** Se detecta un texto inválido y una cantidad negativa. 100 es grande, pero no necesariamente incorrecto.

**Recuerda:** Una cantidad negativa podría ser una devolución. El diccionario de datos debe aclarar si está permitida.
