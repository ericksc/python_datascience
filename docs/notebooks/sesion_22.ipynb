## Pregunta, audiencia y gráfico

Antes de elegir una biblioteca, define qué debe comprender el lector. Comparar categorías, describir una distribución y explorar una relación requieren representaciones distintas. Matplotlib ofrece control fino; Seaborn facilita gráficos estadísticos; Plotly permite exploración interactiva.

Un título con conclusión debe sostenerse en datos y delimitar el contexto. Evita afirmar una tendencia general a partir de un conjunto pequeño o sintético.

```python
import pandas as pd
datos = pd.DataFrame({
    "Sucursal": ["Centro"]*6 + ["Norte"]*6,
    "Producto": ["A", "B", "C"]*4,
    "Unidades": [2, 4, 3, 5, 2, 6, 3, 5, 4, 6, 3, 7],
    "Precio": [100, 200, 150]*4
})
datos["Ingreso"] = datos["Unidades"] * datos["Precio"]
import matplotlib.pyplot as plt
r=datos.groupby("Sucursal").Ingreso.sum()
fig,ax=plt.subplots(figsize=(7,4),layout="constrained")
bars=ax.bar(r.index,r.values,color=["#7155ba","#216556"])
ax.bar_label(bars,fmt="%.0f",padding=3)
ax.set(title="Norte registra más ingreso en estas 12 ventas",ylabel="Ingreso (CRC)",ylim=(0,5000))
plt.show()
```

**Qué observar:** Norte: 4150 CRC; Centro: 3250 CRC. El título limita el hallazgo a las 12 ventas.

**Recuerda:** Mayor ingreso no demuestra mayor rentabilidad ni mejor desempeño causal.

## Accesibilidad y comparaciones honestas

Usa etiquetas directas, contraste legible y símbolos además de color. Para barras, un origen distinto de cero puede exagerar diferencias. En líneas puede ser razonable acotar el eje, pero debes mostrar la escala y justificarla.

Mantén unidades y escalas comparables cuando contrastas grupos. Añade un resumen textual de los hallazgos para personas que no puedan interactuar con el gráfico.

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(figsize=(7,4),layout="constrained")
ax.plot([1,2,3],[10,12,14],marker="o",linestyle="-",label="Centro")
ax.plot([1,2,3],[11,13,15],marker="s",linestyle="--",label="Norte")
ax.set(title="Unidades por día · ejemplo sintético",xlabel="Día",ylabel="Unidades")
ax.legend(title="Sucursal")
plt.show()
```

**Qué observar:** Ambas series se distinguen por etiqueta, marcador y estilo de línea, incluso sin reconocer sus colores.

**Recuerda:** No uses el color como único canal para comunicar una diferencia esencial.

## Una historia verificable

Una entrega puede seguir tres pasos: qué se observó, qué evidencia lo respalda y qué límites condicionan la interpretación. Incluye fuente, período, unidades y reglas de limpieza.

No agregues gráficos que repitan la misma información sin aportar una comparación útil. Selecciona tres vistas complementarias y acompaña cada una con una conclusión breve. Los archivos deben poder abrirse fuera de tu sesión.

```python
import pandas as pd
datos = pd.DataFrame({
    "Sucursal": ["Centro"]*6 + ["Norte"]*6,
    "Producto": ["A", "B", "C"]*4,
    "Unidades": [2, 4, 3, 5, 2, 6, 3, 5, 4, 6, 3, 7],
    "Precio": [100, 200, 150]*4
})
datos["Ingreso"] = datos["Unidades"] * datos["Precio"]
resumen=datos.groupby("Sucursal").Ingreso.sum()
total=resumen.sum()
porcentaje=100*resumen["Norte"]/total
print(f"Norte aporta {porcentaje:.1f}% del ingreso registrado.")
print("Fuente: 12 registros sintéticos; moneda CRC.")
print("Límite: no hay costos ni una muestra representativa del mercado.")
```

**Qué observar:** Norte aporta 56.1 % de 7400 CRC registrados. Se explicitan fuente y límites.

**Recuerda:** Una recomendación debe ser proporcional a lo que la evidencia permite sostener.
