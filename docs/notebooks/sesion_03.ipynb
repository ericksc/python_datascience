## Condiciones: una regla explícita

if ejecuta un bloque cuando una condición es verdadera. elif permite probar otra condición y else cubre el caso restante. La indentación define qué instrucciones pertenecen a cada bloque.

Una regla debe ser precisa: cantidad >= 5 incluye el cinco; cantidad > 5 no lo incluye. Antes de programar, escribe qué debe pasar en el límite.

```python
cantidad = 5
if cantidad >= 5:
    descuento = 0.10
else:
    descuento = 0.0
print("Descuento:", descuento)
```

**Qué observar:** Descuento: 0.1.

**Recuerda:** Prueba valores debajo, en y encima del límite: 4, 5 y 6.

## Ciclos: repetir una operación

for recorre los elementos de una colección. Un acumulador comienza con un valor inicial y se actualiza en cada repetición.

Un ciclo ayuda a comprender el proceso. Más adelante usarás operaciones de Pandas para calcular columnas completas.

```python
cantidades = [4, 10, 3]
total = 0
for cantidad in cantidades:
    total = total + cantidad
print("Unidades:", total)
print("Comprobación:", sum(cantidades))
```

**Qué observar:** Ambos resultados son 17.

**Recuerda:** Inicializa el acumulador antes del ciclo. Si lo reinicias dentro, pierdes lo acumulado.

## Funciones: entradas, resultado y responsabilidad

def define una función. Los parámetros nombran sus entradas. return entrega un resultado que puede usarse en otros cálculos; print solo muestra información.

Una función pequeña debe tener una tarea clara. Aquí calculamos un ingreso con descuento. El descuento se expresa como proporción: 0.10 equivale a 10 %.

```python
def calcular_ingreso(precio, cantidad, descuento=0):
    return precio * cantidad * (1 - descuento)

normal = calcular_ingreso(1500, 4)
rebajado = calcular_ingreso(1500, 4, 0.10)
print(normal, rebajado)
```

**Qué observar:** 6000 y 5400.0.

**Recuerda:** Evita depender de variables externas si puedes pasarlas como parámetros: la función será más fácil de revisar.
