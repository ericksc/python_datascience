## Rutas y archivos de datos

Una ruta relativa se interpreta desde la carpeta de trabajo, que no siempre coincide con la ubicación que imaginas. Path.cwd() permite consultarla.

El paquete contiene las carpetas datos y notebooks. Los ejemplos de lectura de CSV preparan la ruta DATOS buscando los archivos en datos, ../datos o la carpeta actual. En Colab, sube los CSV al panel Archivos.

```python
from pathlib import Path

carpeta = Path.cwd()
print("Carpeta de trabajo:", carpeta.name)
archivo = Path("datos") / "ventas.csv"
print("Ruta relativa:", archivo.as_posix())
```

## CSV: texto organizado en columnas

Un CSV guarda registros en texto separados por un delimitador. Aquí usamos comas. DictReader interpreta la primera fila como nombres de columnas; los valores leídos son texto.

Este ejemplo crea un archivo pequeño de práctica. utf-8 permite guardar acentos. newline="" evita problemas con saltos de línea al escribir CSV.

```python
import csv
from pathlib import Path

ruta = Path("compra_practica.csv")
with ruta.open("w", encoding="utf-8", newline="") as archivo:
    escritor = csv.DictWriter(archivo, fieldnames=["Producto", "Cantidad"])
    escritor.writeheader()
    escritor.writerow({"Producto": "Cuaderno", "Cantidad": 4})

with ruta.open(encoding="utf-8", newline="") as archivo:
    filas = list(csv.DictReader(archivo))
print(filas[0]["Producto"])
print(int(filas[0]["Cantidad"]))
```

**Qué observar:** Cuaderno y 4. Se crea compra_practica.csv en la carpeta de trabajo.

**Recuerda:** La extensión CSV no garantiza separador, codificación o tipos. Inspecciona el archivo antes de interpretarlo.

## Leer un error sin perder el contexto

ValueError suele aparecer cuando una conversión recibe un valor incompatible. KeyError señala una clave inexistente; NameError, un nombre no definido.

Captura errores específicos cuando puedas responder a ellos. Un except general que no informa nada puede esconder problemas. Registra qué no se pudo convertir.

```python
entradas = ["4", "sin dato", "8"]
for texto in entradas:
    try:
        numero = int(texto)
        print(numero)
    except ValueError:
        print("No se pudo convertir:", texto)
```

**Qué observar:** 4; No se pudo convertir: sin dato; 8.

**Recuerda:** Un notebook reproducible incluye importaciones, rutas, decisiones y resultados, y se ejecuta completo después de reiniciar.
