## Forma, transposición y producto

Usa ndarray bidimensional para matrices. * multiplica elemento por elemento; @ realiza producto matricial. Para A de forma (m,n) y B de forma (n,p), A @ B tiene forma (m,p).

reshape reorganiza una cantidad compatible de elementos; .T transpone ejes en una matriz. Una transposición puede ser una vista de los mismos datos: usa copy si necesitas independencia.

```python
import numpy as np
A=np.array([[1,2],[3,4]])
B=np.array([[2,0],[0,2]])
print(A*B)
print(A@B)
print(A.T)
```

**Qué observar:** Elemento a elemento: [[2,0],[0,8]]. Matricial: [[2,4],[6,8]].

**Recuerda:** Comprueba shapes antes de operar. No necesitas np.matrix.

## Broadcasting: compatibilidad, no magia

NumPy compara dimensiones desde la derecha: deben coincidir o una debe ser 1. Una media por columna con keepdims=True tiene forma (1,n) y puede restarse de una matriz (m,n).

Para estandarizar, resta media y divide por desviación. Una columna constante tiene desviación cero: define una política antes de dividir. En modelado, ajusta estas estadísticas solo con entrenamiento.

```python
import numpy as np
X=np.array([[1.,10.],[3.,20.],[5.,30.]])
media=X.mean(axis=0,keepdims=True)
escala=X.std(axis=0,keepdims=True)
Z=(X-media)/np.where(escala==0,1,escala)
print(media.shape)
print(np.round(Z.mean(axis=0),6))
```

**Qué observar:** La media tiene forma (1,2); las medias estandarizadas son [0,0].

**Recuerda:** Broadcasting no implica que las dimensiones tengan significado compatible: verifica unidades y ejes.

## Resolver y verificar

np.linalg.solve encuentra x para A @ x = b si A es cuadrada y de rango completo. Evita calcular la inversa explícitamente solo para resolver el sistema.

Un residuo pequeño confirma que la solución satisface aproximadamente la ecuación. Un sistema mal condicionado puede amplificar errores; revisa cond y considera el contexto. Para un sistema rectangular se estudia lstsq como extensión.

```python
import numpy as np
A=np.array([[2.,1.],[1.,3.]])
b=np.array([5.,10.])
x=np.linalg.solve(A,b)
print(x)
print(np.allclose(A@x,b))
print(np.linalg.norm(A@x-b))
```

**Qué observar:** x=[1,3]; verificación True; residuo cercano a cero.

**Recuerda:** Un residuo pequeño no elimina por sí solo problemas de condicionamiento.
